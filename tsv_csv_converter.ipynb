# Conversion of TSV dataset into pivoted CSV

The Library of Congress archive of published comic book issues comes in a tabular, long format.
The script below helps make the dataset more manageable for further data processing in excel.

```python
import pandas as pd;

df = pd.read_csv('2026-01-10_issues.tsv', sep='\t', header=None);

print(df.head());
print(df[1].unique())
```

The above table is formated...

column 0 - the index of the issue within the table<br>
column 1 - 21 distinct headers describing the issue<br>
column 2 - the factual data<br>

clarifying the pivot below.

```python
# data frame pivot - dfp
dfp = df.pivot(index=0, columns=1, values=2)
```

## Analysis Questions

- How has price varied over time?
- What is the relation between page count and listed price?
- What are the most prolific series and their publishers?

To answer these questions, and to save on memory usage, the next step reduces the dataset to only vital information.

```python
dfp = dfp[dfp["series country code"] == "us"] # only concerned with the US comic market

# data frame pivot reduced - dfpr
dfpr = dfp[["series name", "key date", "publisher name", "brand name", "issue number", "price", "issue page count"]]

print(f"initial size: {df.size}\nnew size: {dfpr.size}\ndelta: {df.size - dfpr.size}")
```

finally exporting the new dataset in CSV format

```python
dfpr.to_csv("2026-01-10_issues_cleaned.csv", index=False)
```
